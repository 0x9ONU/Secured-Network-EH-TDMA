```python
import pandas as pd
import os
import glob
from datetime import datetime
```

```python
def convert_txt_to_csv(input_folder, output_folder):
    """
    Reads .txt files, moves Packet # to the first column, adds a Time Differential (dt),
    and saves them as .csv files.
    """
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    txt_files = glob.glob(os.path.join(input_folder, "*.txt"))

    if not txt_files:
        print("No .txt files found.")
        return

    for file_path in txt_files:
        file_name = os.path.basename(file_path)
        print(f"Processing: {file_name}...")
        
        data_rows = []
        previous_time = None
        
        with open(file_path, 'r') as f:
            for line in f:
                line = line.strip()
                if not line:
                    continue
                
                try:
                    parts = line.split(', ')
                    if len(parts) < 3:
                        continue
                        
                    full_timestamp_str = parts[0]
                    packet_num = parts[1]
                    raw_data = parts[2]
                    
                    # Parse full timestamp for differential calculation
                    current_time = datetime.strptime(full_timestamp_str, '%Y-%m-%d %H:%M:%S.%f')
                    
                    # Calculate delta t (time differential)
                    if previous_time is None:
                        delta_t = 0.0  # First packet has no previous
                    else:
                        delta_t = (current_time - previous_time).total_seconds()
                    
                    previous_time = current_time

                    # Split Date and Time components
                    date_part, time_part = full_timestamp_str.split(' ')
                    hms, partial_seconds = time_part.split('.')
                    hour, minute, second = hms.split(':')
                    
                    # Data segregation (ID:LocalSendTime:SyncFlag:AckFlag:Data)
                    data_parts = raw_data.split(':', 4)
                    while len(data_parts) < 5:
                        data_parts.append("")

                    # Construct row with new order and new column
                    row = {
                        "Packet #": packet_num,
                        "Date": date_part,
                        "Hour": hour,
                        "Minute": minute,
                        "Second": second,
                        "Partial Seconds": partial_seconds,
                        "Time Differential": delta_t,
                        "ID": data_parts[0],
                        "Local Send Time": data_parts[1],
                        "Sync Flag": data_parts[2],
                        "Ack Flag": data_parts[3],
                        "Data": data_parts[4]
                    }
                    data_rows.append(row)
                except Exception as e:
                    print(f"Error on line in {file_name}: {e}")

        # Save to CSV
        df = pd.DataFrame(data_rows)
        csv_file_name = file_name.replace(".txt", ".csv")
        df.to_csv(os.path.join(output_folder, csv_file_name), index=False)
        print(f"Saved: {csv_file_name}")
```

```python
# --- Execution ---
# Replace 'input_data_folder' with the path to your folder containing .txt files
input_folder = '../../../Secured-Network-EH-TDMA/TDMA_Tooling-main/' 
output_folder = 'Converted CSVs'

convert_txt_to_csv(input_folder, output_folder)
```
